Q2. Why does UCS sometimes output the same path but with different costs? (2 marks)

**Answer**:

UCS sometimes output the same path but with different cost because the cost of the steps (nodes visisted) is identical, the cumulative cost g(n) depeneds on the individual action at each steps.


Example:

Path [1, 2, 4, 5, 10] appears in multiple scenarios:

**Scenario 1 (bus cost = 0.5):**
- Actions: bus → bus → walk → bus
- Cost: 0.5 + 0.5 + 1.0 + 0.5 = 2.5

**Scenario 2 (bus cost = 1.0):**
- Actions: bus → bus → walk → bus
- Cost: 1.0 + 1.0 + 1.0 + 1.0 = 4.0

**Scenario 3 (bus cost = 1.5):**
- Actions: walk → bus → walk → bus
- Cost: 1.0 + 1.5 + 1.0 + 1.5 = 5.0

The same state sequence [1, 2, 4, 5, 10] results in different total costs because the action costs (walk = 1.0, bus = variable) change between scenarios. UCS selects the lowest-cost path to the goal, and when the optimal path structure happens to visit the same states, only the total cost differs based on the action costs used.

### Step 0: Initialization

Start with priority queue `pq` where:
* `g` = cumulative cost g(n)
* `s` = current state
* `path` = track visited states

Initialize: `pq = [(0.0, 1, [1])]` and `best = {1: 0.0}`
* State = 1, g(n) = 0 (this corresponds to the starting node in the diagram although i didnt wrote in the diagram :D )

### Step 1: Expansion Process

At each iteration, `g, s, path = heapq.heappop(pq)` removes the state with the lowest cumulative cost from the priority queue.

For example, in Step 1 of the diagram, state 1 is expanded and two successor states are generated:

* **Walk action**: `t = s + 1` with `ng = g + walk_cost`
  - Creates state 2 with cost = 0 + 1 = 1
  
* **Bus action**: `t = 2*s` with `ng = g + bus_cost`
  - Creates state 2 with cost = 0 + 2 = 2

Before adding successors to `pq`, the code checks `if ng < best.get(t, float('inf'))` to ensure only lower-cost paths are explored. If valid, successors are added using `heapq.heappush(pq, (ng, t, path + [t]))`.

### Step 2: Goal Check and Pruning

When a state is popped from `pq`:
1. **Goal check**: `if s == N` checks if the goal is reached. In the diagram, this occurs when state 10 is popped with cost 6.
2. **Pruning**: `if g > best.get(s, float('inf'))` skips states that have already been reached with lower cost, preventing redundant expansions.

The `best` dictionary tracks the minimum cost to reach each state, ensuring UCS always explores the lowest-cost path first and finds the optimal solution.RetryClaude can make mistakes. Please double-check responses.

# UCS problem

````python
# 🔧 Setup
import heapq
import numpy as np
import matplotlib.pyplot as plt
````

````python
# 🚍 UCS Implementation
def ucs_magic_bus(N, walk_cost=1.0, bus_cost=1.0):
    """Uniform Cost Search for Magic Bus problem (start=1, goal=N)."""
    pq = [(0.0, 1, [1])]
    best = {1: 0.0}
    while pq:
        g, s, path = heapq.heappop(pq)
        if s == N:
            return g, path
        if g > best.get(s, float('inf')):
            continue
        if s + 1 <= N:
            t = s + 1
            ng = g + walk_cost
            if ng < best.get(t, float('inf')):
                best[t] = ng
                heapq.heappush(pq, (ng, t, path + [t]))
        if 2*s <= N:
            t = 2*s
            ng = g + bus_cost
            if ng < best.get(t, float('inf')):
                best[t] = ng
                heapq.heappush(pq, (ng, t, path + [t]))
    return None, []
````

````python
# 🎯 Fixed Scenarios
def run_fixed_scenarios(N=10):
    scenarios = [
        ("Bus cheap (< walk)", 0.5),
        ("Bus equal to walk", 1.0),
        ("Bus slightly more expensive", 1.5),
        ("Bus very expensive", 5.0),
        ("Bus negative", -1.0),
    ]
    for desc, cb in scenarios:
        cost, path = ucs_magic_bus(N, walk_cost=1.0, bus_cost=cb)
        print(f"\nScenario: {desc:>26} | bus cost = {cb:>4}")
        print(f"  Path: {path}")
        print(f"  Total cost: {cost}")


# Try with N=10
run_fixed_scenarios(N=10)
````
